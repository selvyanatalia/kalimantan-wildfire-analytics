#### LATAR BELAKANG

Hutan Tropis Kalimantan dikenal dunia sebagai salah satu kawasan hutan hujan tropis tertua dan terbesar, yang menjadi rumah bagi ribuan jenis flora dan fauna langka serta memiliki peran penting dalam menjaga keseimbangan iklim global. Keanekaragaman hayati mencakup lebih dari 15.000 species tumbuhan, lebih dari 600 burung, dan lebih dari 220 mamalia. Namun, dalam beberapa dekade terakhir, habitat hutan kalimantan terus mengalami tekanan akibat penebangan hutan, pembangunan infrastruktur, pertambangan, dan ekspansi perkebunan kelapa sawit. Perubahan tersebut berdampak terhadap hilangnya habitat dan penurunan populasi berbagai jenis flora dan fauna yang bergantung pada ekosistem hutan. Permasalahan tersebut kian relevan saat ini pada quarter 4 tahun 2026 kebakaran hebat melanda hutan hujan tropis dan lahan di wilayah Kalimantan. Kebakaran tidak hanya mengancam kawasan hutan tetapi flora dan fauna yang hidup didalamnya juga terdampak. Oleh karena itu, analisis Kalimantan Wildfire and Ecological Exposure dilakukan untuk mengeksplorasi persebaran aktivitas kebakaran, mengidentifikasi habitat yang terpapar, serta melihat keterkaitannya dengan persebaran keanekaragaman hayati di Kalimantan. Melalui analisis data dan visualisasi dalam dashboard, diharapkan dapat diperoleh gambaran mengenai wilayah yang memiliki paparan ekologis lebih tinggi dan membutuhkan perhatian lebih dalam upaya pemantauan serta perlindungan ekosistem dan satwa liar.

#### SMART FRAMEWORK

**1. Specific** : Menganalisis persebaran aktivitas kebakaran hutan dan lahan di Kalimantan, mengidentifikasi habitat yang terpapar kebakaran, serta mengidentifikasi wilayah yang memiliki potensi risiko ekologis lebih tinggi berdasarkan keberadaan keanekaragaman hayati.

-- Target : Wilayah di Kalimantan yang menjadi titik persebaran hotspot sebagai indikator aktivitas kebakaran hutan dan lahan serta habitat dan keanekaragaman hayati yang berpotensi terdampak.

-- Focus : Menganalisis persebaran hotspot/kebakaran dan mengidentifikasi spesies flora dan fauna yang berstatus terancam berdasarkan habitat yang berpotensi terpapar aktivitas kebakaran. 

-- Tujuan : Mengetahui konsentrasi lokasi/wilayah yang terindikasi hotspot sebagai indikator aktivitas kebakaran, mengidentifikasi jenis flora dan fauna yang berstatus terancam dan habitatnya berpotensi terpapar kebakaran, dan mengidentifikasi wilayah yang memiliki tingkat ecological exporsure / risk lebih tinggi berdasarkan kombinasi titik aktivitas hotspot dan keberadaan biodiversitas. 

**2. Measurable** : Mengetahui persebaran dan tingkat aktivitas kebakaran menggunakan jumlah hotspot, confidence, dan Fire Radiative Power (FRP) sebagai indikator wildfire activity serta species, scientific name, dan jumlah spesies sebagai indikator ecological exposure untuk mengidentifikasi wilayah dengan potensi ecological risk yang lebih tinggi.

-- Metrix Utama : Jumlah titik lokasi hotspot sebagi proxy/indicator utama aktivitas kebakaran hutan dan lahan berdasarkan jumlah deteksi hotspot pada periode pengamatan. 

-- Metrix Pendukung : Confidence label, Fire Radiative Power (FRP) untuk menggambarkan konsentrasi, tingkat kepercayaan deteksi, dan intensitas energi panas yang terdeteksi. 

-- Jumlah Hotspot : Mengukur jumlah deteksi hotspot pada suatu wilayah dan periode tertentu.

-- Confidence Label : Menggambarkan tingkat kepercayaan terhadap deteksi hotspot. 

-- FRP : Menggambarkan energi radiasi panas yang terdeteksi dari fire pixel dan digunakan sebagai indikator pendukung aktivitas/intensitas kebakaran. 

**3. Achievable** : Analisa ini dilakukan dengan mengintegrasikan data hotspot kebakaran dan data biodiversitas untuk melakukan analisis spasial serta mengidentifikasi keterkaitan antara persebaran aktivitas kebakaran dan spesies flora dan fauna berstatus terancam. 

-- Variabel Utama : Jumlah hotspot digunakan sebagai variabel utama untuk mengukur aktivitas kebakaran selama periode pengamatan.

-- Variabel Kebakaran : Latitude, longitude, tanggal, waktu, confidence label, dan Fire Radiative Power (FRP) digunakan untuk menganalisis persebaran, waktu, tingkat kepercayaan, dan karakteristik aktivitas hotspot.

-- Variabel Habitat : Tipe habitat atau land cover digunakan untuk mengidentifikasi jenis habitat yang berada pada area yang terpapar atau berdekatan dengan aktivitas hotspot.

-- Variabel Biodiversitas : Nama spesies, kelompok taksonomi, habitat digunakan untuk mengidentifikasi flora dan fauna yang berada pada habitat yang berpotensi terpapar aktivitas kebakaran, terutama spesies yang berstatus terancam.

-- Variabel Kategorikal : Tipe habitat, kelompok flora/fauna digunakan untuk melakukan segmentasi dan membandingkan karakteristik biodiversitas pada habitat yang berpotensi terpapar.

-- Spatial Analysis : Data latitude dan longitude hotspot akan digunakan untuk mengidentifikasi lokasi aktivitas kebakaran dan menghubungkannya dengan data habitat untuk mengetahui jenis habitat yang berpotensi terpapar. Data biodiversitas kemudian digunakan untuk mengidentifikasi spesies yang tercatat pada habitat tersebut.

-- Ecological Exposure/Risk Analysis : Ecological exposure/risk dianalisis berdasarkan kombinasi aktivitas hotspot, habitat yang terpapar, serta keberadaan spesies flora dan fauna, dengan perhatian khusus terhadap spesies yang memiliki status konservasi terancam. 

**4. Relevant** : Hasil analisis ini relevan untuk memberikan gambaran berbasis data mengenai persebaran aktivitas kebakaran serta potensi paparan terhadap habitat dan spesies flora dan fauna berstatus terancam di Kalimantan, terutama dalam konteks kejadian kebakaran hutan dan lahan pada Quarter 4 Tahun 2026.

-- Issue : Kebakaran hutan dan lahan tidak hanya berkaitan dengan persebaran titik panas, tetapi juga berpotensi mengganggu habitat dan kehidupan satwa liar. Beberapa kejadian pada Quarter 4 Tahun 2026 menunjukkan adanya satwa liar yang ditemukan keluar dari habitatnya maupun ditemukan dalam kondisi terdampak di wilayah yang mengalami karhutla.

-- Analytical Gap : Informasi mengenai kebakaran umumnya menunjukkan lokasi dan jumlah hotspot, tetapi belum memberikan gambaran terintegrasi mengenai habitat apa yang berpotensi terpapar dan spesies berstatus terancam apa yang berada pada habitat tersebut.

**5. Time Bound** : Analisa ini dilakukan dengan focus pada aktivitas kebakaran hutan dan lahan di Kalimantan pada periode Quarter 4 Tahun 2026 dan menghubungkannya dengan data habitat serta biodiversitas untuk menggambarkan potensi ecological exposure pada periode analisis. 

-- Periode pengamatan : Quarter 4 Tahun 2026 dengan focus data hotspot yang tersedia selama periode pengamatan (NASA VIIRS).

-- Biodiversitas dan Habitat : Data Flora dan Fauna menggunakan data real occurrence dari GBIF (Global Biodiversity Information Facility) dengan wilayah observasi keseluruhan Pulau Kalimantan.


#### FINAL PROBLEM STATEMENT 

Aktivitas kebakaran hutan dan lahan di Kalimantan berpotensi mengancam habitat dan keanekaragaman hayati, namun belum terdapat gambaran yang terintegrasi mengenai wilayah yang memiliki paparan risiko ekologis tinggi berdasarkan persebaran hotspot, intensitas kebakaran, habitat dan keberadaan species terancam. Hal ini dapat menyulitkan identifikasi wilayah yang perlu menjadi prioritas dalam perlindungan ekosistem dan keanekaragaman hayati (biodiversitas).
